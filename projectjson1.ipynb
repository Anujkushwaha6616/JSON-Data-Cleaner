```python
import json 
```

```python
def load_data(file_name):
    with open(file_name,"r")as file:
        data = json.load(file)
    return data
```

```python
data = load_data("anuj1.json")
```

```python
def display_data(data):
    for user in data["users"]:
        print("*______________USER____________*")
        print(f"User id: {user['id']}\t Name:  {user['name']}\tFriends :{user['friends']}\tLike pages: {user['liked_pages']}")
    for page in data["pages"]:
        print("*______________PAGE____________*")
        print(f"Page id: {page['id']}\t Page name: {page['name']}")
```

```python
display_data(data)
```

```python
def cleaned_data(data):
    #remove without name user
    data['users'] = [user for user in data['users'] if user.get('name') and user['name'].strip()]

    #remove duplicate friends
    for user in data['users']:
        user['friends'] = list(set(user['friends']))

    #remove inactive uer
    data['users'] = [user for user in data['users'] if user['friends'] or user['liked_pages']]


    #remove duplicate pages
    unique_pages = {}
    for page in data['pages']:
        unique_pages[page['id']] = page
    data['pages'] = list(unique_pages.values())
    return data

    


```

```python
data = json.load(open("anuj1.json"))
prepared_data = cleaned_data(data)

json.dump(prepared_data, open("cleared_data.json", "w"),indent=4)
print("work Done successfully!")
```
